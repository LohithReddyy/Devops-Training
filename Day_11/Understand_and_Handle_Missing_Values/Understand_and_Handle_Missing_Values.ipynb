```python
df5 = df.head(5).copy()

df5.dropna()
```

```python
df5["UnitPrice"] = pd.to_numeric(
    df5["UnitPrice"], errors="coerce"
)

print("Mean:", df5["UnitPrice"].mean())
print("Median:", df5["UnitPrice"].median())
```

```python
df5["Quantity"] = pd.to_numeric(
    df5["Quantity"], errors="coerce"
)

df5["Quantity"] = df5["Quantity"].fillna(
    df5["Quantity"].median()
)

df5
```

```python
df5["UnitPrice"] = df5["UnitPrice"].fillna(
    df5["UnitPrice"].median()
)

df5
```
