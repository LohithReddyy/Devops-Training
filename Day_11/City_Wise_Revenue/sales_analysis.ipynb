```python
df["Revenue"] = df["Quantity"] * df["UnitPrice"]
```

```python
df["City"] = df["City"].astype("string").str.strip().str.title()
```

```python
city_revenue = df.groupby("City")["Revenue"].sum(min_count=1)
```

```python
city_revenue.sort_values(ascending=False)
```

```python
city_revenue.sort_values(ascending=False)
```

```python
city_summary = (
    df.groupby("City", dropna=False)
      .agg(
          TotalRevenue=("Revenue", "sum"),
          TotalQuantity=("Quantity", "sum"),
          NumberOfOrders=("OrderID", "nunique")
      )
      .sort_values("TotalRevenue", ascending=False)
)

city_summary
```

```python
df = df.dropna(subset=["Revenue"])
```

```python
df.to_csv("sales_revenue_analysis.csv", index=False)
```
