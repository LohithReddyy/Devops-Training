```python
df = df.drop_duplicates()
```

```python
df.duplicated().sum()
```

```python
df["Product"] = df["Product"].str.strip().str.title()

df["City"] = df["City"].str.strip().str.title()
```

```python
print(df["Product"].unique())
print(df["City"].unique())
```

```python
df["OrderDate"] = pd.to_datetime(
    df["OrderDate"],
    errors="coerce"
)
```

```python
df[df["OrderDate"].isna()]
```

```python
df["Quantity"] = pd.to_numeric(
    df["Quantity"],
    errors="coerce"
)

df["UnitPrice"] = pd.to_numeric(
    df["UnitPrice"],
    errors="coerce"
)
```

```python
df.dtypes
```

```python
df["Quantity"] = df["Quantity"].fillna(
    df["Quantity"].median()
)

df["UnitPrice"] = df["UnitPrice"].fillna(
    df["UnitPrice"].median()
)
```

```python
df["City"] = df["City"].fillna("Unknown")

df["Product"] = df["Product"].fillna("Unknown")
```

```python
def price_category(price):
    if pd.isna(price):
        return "Unknown"
    elif price < 100:
        return "Low"
    elif price < 500:
        return "Medium"
    else:
        return "High"
```

```python
df["PriceCategory"] = df["UnitPrice"].apply(price_category)
```

```python
df[["Product", "UnitPrice", "PriceCategory"]]
```

```python
df.nlargest(5, "UnitPrice")[
    ["OrderID", "Product", "UnitPrice", "City"]
]
```

```python
Q1 = df["UnitPrice"].quantile(0.25)
Q3 = df["UnitPrice"].quantile(0.75)

IQR = Q3 - Q1

upper_bound = Q3 + 1.5 * IQR

df[df["UnitPrice"] > upper_bound][
    ["OrderID", "Product", "UnitPrice", "City"]
]
```

```python
df["Revenue"] = df["Quantity"] * df["UnitPrice"]
```
